## **Read the CSV file**

```python
import pyspark.sql.functions as func
from pyspark.sql.types import *


df_sales_extended = spark.read.csv(
    '/Volumes/workspace/default/raw_data/data_sales_extended.csv',
    header=True
)

df_sales_extended.printSchema()
display(df_sales_extended)
```

### **Change types of columns**

```python
df_transformed = df_sales_extended.select(
    func.col('transactionID').try_cast(LongType()),
    func.col('customerID').try_cast(LongType()),
    func.col('franchiseID').try_cast(LongType()),
    func.col('dateTime').try_cast(TimestampType()),
    func.col('product').try_cast(StringType()),
    func.col('quantity').try_cast(LongType()),
    func.col('unitPrice').try_cast(LongType()),
    func.col('totalPrice').try_cast(LongType()),
    func.col('paymentMethod').try_cast(StringType()),
    func.col('cardNumber').try_cast(LongType())
).dropna()
df_transformed.printSchema()
display(df_transformed)
```

## **Create a Raw Database and a Table**

```python
# Create a database and a table
spark.sql(
    '''
        CREATE DATABASE IF NOT EXISTS sales_db_raw
    '''
)

spark.sql(
    '''
        CREATE TABLE IF NOT EXISTS sales_db_raw.sales_extended
        (   transactionID LONG,
            customerID LONG,
            franchiseID LONG,
            dateTime TIMESTAMP,
            product STRING,
            quantity LONG,
            unitPrice LONG,
            totalPrice LONG,
            paymentMethod STRING,
            cardNumber LONG)
    '''
)
```

## **Load the data**

```python
df_transformed.write.mode('overwrite').saveAsTable('sales_db_raw.sales_extended')
```
